```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df_hk_nov= pd.read_csv("linkedin_jobs_20251201_hk_final.csv", on_bad_lines='skip') #scrapped with legacy code, different format (missing ID, seperate keywords)
df_uk_nov= pd.read_csv("linkedin_jobs_20251201_uk_final.csv",engine='python', on_bad_lines='skip')

df_hk_dec= pd.read_csv("linkedin_jobs_20260101_hk_final.csv", on_bad_lines='skip')
df_uk_dec= pd.read_csv("linkedin_jobs_20260101_uk_final.csv", engine='python', on_bad_lines='skip')
df_part_dec=pd.read_csv("linkedin_jobs_20260101_both_345.csv", on_bad_lines='skip')

df_hk_jan= pd.read_csv("linkedin_jobs_20260201_hk_final.csv", on_bad_lines='skip')
df_uk_jan= pd.read_csv("linkedin_jobs_20260201_uk_final.csv", engine='python', on_bad_lines='skip')

df_hk_feb= pd.read_csv("linkedin_jobs_20260301_hk_final.csv", on_bad_lines='skip')
df_uk_feb= pd.read_csv("linkedin_jobs_20260301_uk_final.csv", engine='python', on_bad_lines='skip')

df_nov = pd.concat([df_hk_nov, df_uk_nov], ignore_index=True)
df_dec = pd.concat([df_hk_dec, df_uk_dec,df_part_dec], ignore_index=True)
df_jan = pd.concat([df_hk_jan, df_uk_jan], ignore_index=True)
df_feb = pd.concat([df_hk_feb, df_uk_feb], ignore_index=True)
```

```python
# extract ID (assigned by LinkedIn) from url
df_nov["job_id"] = df_nov["url"].str.extract(r"currentJobId=(\d+)")
```

```python
# add month column
df_nov["month"] = "nov"
df_dec["month"] = "dec"
df_jan["month"] = "jan"
df_feb["month"] = "feb"
```

```python
# append keyword if dup exists then drop
df_nov = (
    df_nov.groupby("job_id").agg({
        "keyword": lambda x: ", ".join(set(x)),
        "location": "first",
        "month": "first",
        "title": "first",
        "company": "first",
        "detail": "first",
        "description": "first"
    })
    .reset_index()
)
```

```python
# append keyword if dup exists then drop (dec)
df_dec = (
    df_dec.groupby("job_id").agg({
        "keyword": lambda x: ", ".join(set(x)),
        "location": "first",
        "month": "first",
        "title": "first",
        "company": "first",
        "detail": "first",
        "description": "first"
    })
    .reset_index())
```

```python
# append keyword if dup exists then drop (jan)
df_jan = (
    df_jan.groupby("job_id").agg({
        "keyword": lambda x: ", ".join(set(x)),
        "location": "first",
        "month": "first",
        "title": "first",
        "company": "first",
        "detail": "first",
        "description": "first"
    })
    .reset_index())
```

```python
# append keyword if dup exists then drop (feb)
df_feb = (
    df_feb.groupby("job_id").agg({
        "keyword": lambda x: ", ".join(set(x)),
        "location": "first",
        "month": "first",
        "title": "first",
        "company": "first",
        "detail": "first",
        "description": "first"
    })
    .reset_index())
```

```python
# drop dup values
df_nov = df_nov.dropna()
df_dec = df_dec.dropna()
```

```python
df = pd.concat([df_nov, df_dec, df_jan, df_feb], ignore_index=True)
```

```python
df.info()
```

```python
df = df[df["detail"].str.contains("Full-time", na=False)].copy()

df.loc[:, "detail"] = (
    df["detail"]
      .str.replace(r"\d+ of \d+ skills match", "", regex=True)
      .str.replace("Matches your job preferences, job type is {}.", "", regex=False)
      .str.replace("\n", " ")
)

df
```

```python
print(f"Number of Jobs in Hong Kong: {len(df[df["location"]=="Hong Kong"])}")
print(f"Number of Jobs in London: {len(df[df["location"]=="London"])}")
```

```python
# save as parquet
df["job_id"] = df["job_id"].astype(str)
df.to_csv("scraped_jobs.csv", index=False)
df.to_parquet("scraped_jobs.parquet", index=False)
```
